```python
import cv2
import os
import matplotlib.pyplot as plt
from pathlib import Path

project_dir = Path.cwd()
if not (project_dir / 'foto_citra').is_dir():
    project_dir = project_dir.parent

input_folder = project_dir / 'foto_citra'
output_folder = project_dir / 'tugas3' / 'hasil_enhancement'

if not input_folder.is_dir():
    raise FileNotFoundError(f'Folder input tidak ditemukan: {input_folder}')

output_folder.mkdir(parents=True, exist_ok=True)

files_list = [
    '01_HighQuality_Enhanced.jpg',
    '02_LowContrast.jpg',
    '03_Blurred.jpg',
    '04_HighNoise.jpg',
    '05_LowResolution_Upsampled.jpg',
    '06_Faded_Underexposed.jpg',
    '07_ColorShift_WarmTint.jpg',
    '08_JPEGCompression_Artifacts.jpg',
    '09_CombinedDegradation.jpg'
]

x1, x2 = 200, 1000
y1, y2 = 2200, 2400

for filename in files_list:

    path = os.path.join(input_folder, filename)
    img = cv2.imread(path)

    if img is None:
        print(f'Gagal membaca: {filename}')
        continue

    # Rotasi 90 derajat clockwise
    img_rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

    # Crop area nomor ijazah
    crop = img_rotated[y1:y2, x1:x2]

    # Grayscale
    gray = cv2.cvtColor(crop, cv2.COLOR_BGR2GRAY)

    # 1. Brightness adjustment
    brightness = cv2.convertScaleAbs(gray, alpha=1.0, beta=40)

    # 2. Contrast stretching
    contrast = cv2.normalize(
        gray,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    # 3. Histogram equalization
    equalized = cv2.equalizeHist(gray)

    # Nama tanpa extension
    name = os.path.splitext(filename)[0]

    # Simpan hasil
    cv2.imwrite(
        os.path.join(output_folder, f'{name}_crop_gray.jpg'),
        gray
    )

    cv2.imwrite(
        os.path.join(output_folder, f'{name}_brightness.jpg'),
        brightness
    )

    cv2.imwrite(
        os.path.join(output_folder, f'{name}_contrast_stretching.jpg'),
        contrast
    )

    cv2.imwrite(
        os.path.join(output_folder, f'{name}_histogram_equalization.jpg'),
        equalized
    )

    # =========================
    # TAMPILKAN CITRA
    # =========================

    plt.figure(figsize=(16, 4))

    plt.subplot(1, 4, 1)
    plt.imshow(gray, cmap='gray')
    plt.title('Sebelum Enhancement')
    plt.axis('off')

    plt.subplot(1, 4, 2)
    plt.imshow(brightness, cmap='gray')
    plt.title('Brightness Adjustment')
    plt.axis('off')

    plt.subplot(1, 4, 3)
    plt.imshow(contrast, cmap='gray')
    plt.title('Contrast Stretching')
    plt.axis('off')

    plt.subplot(1, 4, 4)
    plt.imshow(equalized, cmap='gray')
    plt.title('Histogram Equalization')
    plt.axis('off')

    plt.suptitle(name)
    plt.tight_layout()

    plt.savefig(
        os.path.join(output_folder, f'{name}_perbandingan.jpg'),
        dpi=150
    )

    plt.show()

    # =========================
    # TAMPILKAN HISTOGRAM
    # =========================

    plt.figure(figsize=(16, 4))

    plt.subplot(1, 4, 1)
    plt.hist(gray.ravel(), bins=256, range=[0, 256])
    plt.title('Histogram Sebelum')
    plt.xlabel('Intensitas')
    plt.ylabel('Jumlah Piksel')

    plt.subplot(1, 4, 2)
    plt.hist(brightness.ravel(), bins=256, range=[0, 256])
    plt.title('Histogram Brightness')
    plt.xlabel('Intensitas')
    plt.ylabel('Jumlah Piksel')

    plt.subplot(1, 4, 3)
    plt.hist(contrast.ravel(), bins=256, range=[0, 256])
    plt.title('Histogram Contrast Stretching')
    plt.xlabel('Intensitas')
    plt.ylabel('Jumlah Piksel')

    plt.subplot(1, 4, 4)
    plt.hist(equalized.ravel(), bins=256, range=[0, 256])
    plt.title('Histogram Equalization')
    plt.xlabel('Intensitas')
    plt.ylabel('Jumlah Piksel')

    plt.suptitle(f'Histogram - {name}')
    plt.tight_layout()

    plt.savefig(
        os.path.join(output_folder, f'{name}_histogram.jpg'),
        dpi=150
    )

    plt.show()

print('Selesai. Semua hasil tersimpan di:')
print(output_folder)
```
