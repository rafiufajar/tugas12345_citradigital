```python
import cv2
import matplotlib.pyplot as plt
from pathlib import Path

project_dir = Path.cwd()
if not (project_dir / 'foto_citra').is_dir():
    project_dir = project_dir.parent

input_folder = project_dir / 'foto_citra'

if not input_folder.is_dir():
    raise FileNotFoundError(f'Folder input tidak ditemukan: {input_folder}')

ijazah_mentah = [
    '01_HighQuality_Enhanced.jpg',
    '02_LowContrast.jpg',
    '03_Blurred.jpg',
    '04_HighNoise.jpg',
    '05_LowResolution_Upsampled.jpg',
    '06_Faded_Underexposed.jpg',
    '07_ColorShift_WarmTint.jpg',
    '08_JPEGCompression_Artifacts.jpg',
    '09_CombinedDegradation.jpg'
]

for nama in ijazah_mentah:
    path = input_folder / nama
    img = cv2.imread(str(path))
    if img is None:
        print(f'Gagal membaca: {path}')
        continue

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    plt.figure(figsize=(15, 4))

    # Citra asli
    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title('Citra Asli')
    plt.axis('off')

    # Citra grayscale
    plt.subplot(1, 3, 2)
    plt.imshow(gray, cmap='gray')
    plt.title('Grayscale')
    plt.axis('off')

    # Histogram grayscale
    plt.subplot(1, 3, 3)
    plt.hist(gray.ravel(), bins=256, range=[0, 256])
    plt.title('Histogram Grayscale')
    plt.xlabel('Intensitas')
    plt.ylabel('Jumlah Piksel')

    plt.suptitle(nama)
    plt.show()
```
